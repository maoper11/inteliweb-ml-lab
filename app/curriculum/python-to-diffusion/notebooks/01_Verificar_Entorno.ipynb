# 01 · Verificar el entorno

**Objetivo:** comprobar Python, paquetes, PyTorch, CUDA, GPU y rutas de trabajo.

> Ejecuta todas las celdas. Esta lección no cambia la configuración del sistema.

```python
import sys
from pathlib import Path

print("Python:", sys.version.split()[0])
print("Ejecutable:", sys.executable)
print("Carpeta actual:", Path.cwd())
```

## 📖 ¿Qué debemos comprobar?

Un entorno funcional necesita:

- El Python aislado de `app/env`.
- JupyterLab ejecutándose con ese Python.
- PyTorch importable.
- CUDA disponible cuando existe una GPU NVIDIA compatible.

```python
import importlib

packages = ["numpy", "pandas", "sklearn", "torch", "torchvision", "PIL", "cv2"]

for package in packages:
    module = importlib.import_module(package)
    version = getattr(module, "__version__", "sin versión visible")
    print(f"{package:12} {version}")
```

```python
import torch

print("PyTorch:", torch.__version__)
print("CUDA disponible:", torch.cuda.is_available())
print("CUDA runtime:", torch.version.cuda)

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
    print("Cantidad de GPU:", torch.cuda.device_count())
else:
    print("Se usará CPU en los ejemplos compatibles.")
```

## 🧪 Práctica rápida

Predice el resultado de la siguiente celda antes de ejecutarla.

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
print("Dispositivo seleccionado:", device)
```

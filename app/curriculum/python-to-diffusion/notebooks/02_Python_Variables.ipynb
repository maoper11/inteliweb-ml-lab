# 02 · Python: variables y tipos

**Objetivo:** usar números, texto, booleanos, conversiones y expresiones.

Las variables guardan valores que pueden reutilizarse y modificarse.

```python
modelo = "Krea 2"
steps = 20
cfg = 4.5
usar_gpu = True

print(modelo)
print(steps)
print(cfg)
print(usar_gpu)
```

```python
print(type(modelo))
print(type(steps))
print(type(cfg))
print(type(usar_gpu))
```

## 📖 Operaciones

Python permite combinar valores. Los operadores más comunes son `+`, `-`, `*`, `/`, `**`.

```python
ancho = 1024
alto = 1024
pixeles = ancho * alto

print("Resolución:", ancho, "x", alto)
print("Píxeles:", pixeles)
print("Megapíxeles:", pixeles / 1_000_000)
```

## ✏️ Cadenas formateadas

Las *f-strings* insertan variables dentro de texto.

```python
seed = 12345
print(f"Modelo: {modelo} | Steps: {steps} | CFG: {cfg:.2f} | Seed: {seed}")
```

## 🧪 Práctica

Cambia los valores y calcula cuántas imágenes se producen.

```python
batch_size = 2
cantidad_batches = 3
total_imagenes = batch_size * cantidad_batches

print("Total de imágenes:", total_imagenes)
```

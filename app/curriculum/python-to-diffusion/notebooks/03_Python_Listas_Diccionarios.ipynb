# 03 · Python: listas y diccionarios

**Objetivo:** organizar colecciones de parámetros y recorrerlas con bucles.

```python
loras = ["character.safetensors", "style.safetensors", "lighting.safetensors"]
print(loras)
print("Primera LoRA:", loras[0])
print("Cantidad:", len(loras))
```

```python
for lora in loras:
    print("Cargar:", lora)
```

## 📖 Diccionarios

Un diccionario relaciona claves con valores.

```python
sampler = {"name": "euler", "steps": 20, "cfg": 4.5, "seed": 12345}
print(sampler["name"])
print(sampler["steps"])
```

```python
loras_config = [
    {"name": "character.safetensors", "strength": 1.0, "enabled": True},
    {"name": "style.safetensors", "strength": 0.65, "enabled": False},
]

for item in loras_config:
    if item["enabled"]:
        print(f'Activa: {item["name"]} | strength={item["strength"]:.2f}')
```

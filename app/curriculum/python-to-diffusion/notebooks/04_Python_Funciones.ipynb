**Objetivo:** encapsular tareas reutilizables, recibir parámetros y devolver resultados.

```python
def saludar(nombre):
    return f"Hola, {nombre}"

print(saludar("Mauricio"))
```

## 📖 Parámetros predeterminados

```python
def describir_generacion(modelo, steps=20, cfg=4.5):
    return f"{modelo}: {steps} steps, CFG {cfg:.2f}"

print(describir_generacion("Krea 2"))
print(describir_generacion("Flux", steps=28, cfg=3.5))
```

```python
def calcular_megapixeles(ancho, alto):
    return (ancho * alto) / 1_000_000

print(calcular_megapixeles(1024, 1024))
```

## 🧪 Práctica

Limita una fuerza LoRA al intervalo de `0.0` a `2.0`.

```python
def limitar_strength(valor):
    return max(0.0, min(2.0, valor))

for valor in [-0.5, 0.8, 2.7]:
    print(valor, "->", limitar_strength(valor))
```

# 05 · NumPy: arrays y operaciones

**Objetivo:** comprender arrays, dimensiones, índices, forma y operaciones vectorizadas.

```python
import numpy as np

valores = np.array([10, 20, 30, 40])
print(valores)
print("Forma:", valores.shape)
print("Tipo:", valores.dtype)
```

```python
matriz = np.array([[1, 2, 3], [4, 5, 6]])
print(matriz)
print("Dimensiones:", matriz.ndim)
print("Forma:", matriz.shape)
print("Elemento fila 1, columna 2:", matriz[1, 2])
```

## 📖 Operaciones vectorizadas

No necesitamos un bucle para transformar cada elemento.

```python
normalizado = valores / 40
print(normalizado)
print("Promedio:", normalizado.mean())
```

```python
imagen_gris = np.arange(25).reshape(5, 5)
print(imagen_gris)
print("Zona central:\n", imagen_gris[1:4, 1:4])
```

## 🧪 Práctica

Cambia el brillo multiplicando por otro valor.

```python
pixeles = np.array([20, 80, 160, 240], dtype=np.float32)
brillo = 1.25
resultado = np.clip(pixeles * brillo, 0, 255).astype(np.uint8)
print(resultado)
```

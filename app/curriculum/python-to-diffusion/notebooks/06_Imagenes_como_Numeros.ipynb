# 06 · Imágenes como números

**Objetivo:** crear, inspeccionar y modificar una imagen RGB como un array NumPy.

```python
import numpy as np
import matplotlib.pyplot as plt

alto, ancho = 256, 256
imagen = np.zeros((alto, ancho, 3), dtype=np.uint8)
print("Forma:", imagen.shape)
print("Tipo:", imagen.dtype)
```

```python
imagen[:, :, 0] = 220
imagen[:, :, 1] = 80
imagen[:, :, 2] = 120
plt.imshow(imagen)
plt.axis("off")
plt.show()
```

## 📖 Canales RGB

Cada píxel tiene tres valores: rojo, verde y azul, normalmente entre `0` y `255`.

```python
gradiente = np.zeros((256, 256, 3), dtype=np.uint8)
gradiente[:, :, 0] = np.linspace(0, 255, 256, dtype=np.uint8)
gradiente[:, :, 1] = 100
gradiente[:, :, 2] = 200
plt.imshow(gradiente)
plt.axis("off")
plt.show()
```

```python
invertida = 255 - gradiente
plt.imshow(invertida)
plt.axis("off")
plt.show()
```

## 🧪 Práctica

Crea una franja blanca en el centro de la imagen.

```python
editada = gradiente.copy()
editada[100:156, :, :] = 255
plt.imshow(editada)
plt.axis("off")
plt.show()
```

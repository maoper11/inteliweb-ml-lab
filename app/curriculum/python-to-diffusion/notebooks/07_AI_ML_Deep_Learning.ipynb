# 07 · IA, Machine Learning y Deep Learning

**Objetivo:** distinguir reglas, aprendizaje automático, redes neuronales e IA generativa.

## 📖 Mapa rápido

- **IA:** campo amplio que busca que las máquinas realicen tareas asociadas con inteligencia.
- **Machine Learning:** aprende patrones desde datos.
- **Deep Learning:** usa redes neuronales con varias capas.
- **IA generativa:** produce contenido nuevo, como texto, imágenes, audio o video.

```python
def sistema_reglas(temperatura):
    if temperatura > 30:
        return "encender ventilador"
    return "mantener apagado"

for valor in [22, 35]:
    print(valor, sistema_reglas(valor))
```

El sistema anterior no aprende: alguien escribió la regla. Ahora veremos un modelo que ajusta una relación desde ejemplos.

```python
import numpy as np
from sklearn.linear_model import LinearRegression

horas = np.array([[1], [2], [3], [4], [5]])
resultados = np.array([52, 58, 66, 73, 81])
modelo = LinearRegression()
modelo.fit(horas, resultados)
print("Predicción para 6 horas:", modelo.predict([[6]])[0])
```

## 🧪 Clasifica cada caso

1. Un `if` que selecciona un sampler.
2. Un clasificador entrenado con imágenes.
3. Un modelo de difusión que genera una foto.

```python
casos = {
    "regla": "programación tradicional",
    "clasificador": "machine learning / deep learning",
    "difusión": "IA generativa y deep learning",
}
for caso, categoria in casos.items():
    print(caso, "->", categoria)
```

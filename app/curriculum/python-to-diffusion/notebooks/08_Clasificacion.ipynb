# 08 · Clasificación con Scikit-learn

**Objetivo:** entrenar, evaluar y usar un modelo que asigna categorías.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix

data = load_iris()
print(data.feature_names)
print(data.target_names)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    data.data, data.target, test_size=0.25, random_state=42, stratify=data.target
)
print("Entrenamiento:", X_train.shape)
print("Prueba:", X_test.shape)
```

```python
model = LogisticRegression(max_iter=300)
model.fit(X_train, y_train)
pred = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, pred))
print("Matriz de confusión:\n", confusion_matrix(y_test, pred))
```

## 📖 Inferencia

Después del entrenamiento, `predict` ejecuta inferencia sobre datos nuevos.

```python
muestra = [[5.1, 3.5, 1.4, 0.2]]
clase = model.predict(muestra)[0]
print("Clase predicha:", data.target_names[clase])
```

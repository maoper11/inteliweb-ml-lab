# 09 · Perceptrón desde cero

**Objetivo:** entender pesos, bias, suma ponderada y función de activación.

```python
import numpy as np

def escalon(valor):
    return 1 if valor >= 0 else 0

def perceptron(x, pesos, bias):
    suma = np.dot(x, pesos) + bias
    return escalon(suma), suma
```

## 📖 Compuerta AND

Solo devuelve `1` cuando ambas entradas son `1`.

```python
pesos = np.array([1.0, 1.0])
bias = -1.5
for x1 in [0, 1]:
    for x2 in [0, 1]:
        salida, suma = perceptron([x1, x2], pesos, bias)
        print([x1, x2], "suma=", suma, "salida=", salida)
```

```python
pesos_or = np.array([1.0, 1.0])
bias_or = -0.5
for x1 in [0, 1]:
    for x2 in [0, 1]:
        salida, _ = perceptron([x1, x2], pesos_or, bias_or)
        print([x1, x2], "->", salida)
```

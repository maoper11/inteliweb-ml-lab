# 10 · Tensores en PyTorch

**Objetivo:** crear tensores, inspeccionar formas, cambiar tipos y usar CPU o GPU.

```python
import torch

tensor = torch.tensor([1.0, 2.0, 3.0])
print(tensor)
print("Forma:", tensor.shape)
print("dtype:", tensor.dtype)
print("device:", tensor.device)
```

```python
matriz = torch.arange(12, dtype=torch.float32).reshape(3, 4)
print(matriz)
print("Promedio:", matriz.mean())
print("Columna 2:", matriz[:, 2])
```

## 📖 Broadcasting

```python
imagen = torch.zeros(3, 4, 4)
offset_rgb = torch.tensor([0.1, 0.2, 0.3]).view(3, 1, 1)
resultado = imagen + offset_rgb
print(resultado.shape)
print(resultado[:, 0, 0])
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
matriz_device = matriz.to(device)
print("Dispositivo:", matriz_device.device)
```

## 🧪 Práctica

Compara el espacio usado por FP32 y FP16.

```python
a = torch.zeros(1000, dtype=torch.float32)
b = torch.zeros(1000, dtype=torch.float16)
print("FP32 bytes:", a.element_size() * a.numel())
print("FP16 bytes:", b.element_size() * b.numel())
```

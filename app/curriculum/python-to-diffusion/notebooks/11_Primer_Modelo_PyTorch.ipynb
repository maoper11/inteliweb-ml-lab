# 11 · Primer modelo con PyTorch

**Objetivo:** definir una red, entrenarla y observar cómo disminuye la pérdida.

```python
import torch
from torch import nn

torch.manual_seed(42)
X = torch.linspace(-1, 1, 100).unsqueeze(1)
y = 3 * X + 0.5 + 0.1 * torch.randn_like(X)
print(X.shape, y.shape)
```

```python
class ModeloLineal(nn.Module):
    def __init__(self):
        super().__init__()
        self.capa = nn.Linear(1, 1)

    def forward(self, x):
        return self.capa(x)

modelo = ModeloLineal()
print(modelo)
```

```python
loss_fn = nn.MSELoss()
optimizer = torch.optim.SGD(modelo.parameters(), lr=0.1)

for epoch in range(101):
    pred = modelo(X)
    loss = loss_fn(pred, y)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    if epoch % 20 == 0:
        print(f"epoch={epoch:3} loss={loss.item():.6f}")
```

```python
with torch.no_grad():
    prediccion = modelo(torch.tensor([[2.0]]))
print("Predicción para x=2:", prediccion.item())
print("Peso aprendido:", modelo.capa.weight.item())
print("Bias aprendido:", modelo.capa.bias.item())
```

## 🧪 Práctica

Cambia el learning rate de `0.1` a `0.01`, reinicia el kernel y compara.

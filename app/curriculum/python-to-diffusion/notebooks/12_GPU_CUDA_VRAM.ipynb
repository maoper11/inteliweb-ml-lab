# 12 · GPU, CUDA y VRAM

**Objetivo:** identificar el dispositivo, medir memoria y entender `allocated` frente a `reserved`.

```python
import torch
print("PyTorch:", torch.__version__)
print("CUDA disponible:", torch.cuda.is_available())
print("CUDA runtime:", torch.version.cuda)
```

```python
if torch.cuda.is_available():
    index = torch.cuda.current_device()
    props = torch.cuda.get_device_properties(index)
    print("GPU:", props.name)
    print(f"VRAM total: {props.total_memory / 1024**3:.2f} GB")
else:
    print("No hay CUDA activa.")
```

```python
def mostrar_memoria():
    if not torch.cuda.is_available():
        print("CUDA no disponible")
        return
    libre, total = torch.cuda.mem_get_info()
    print(f"Libre: {libre / 1024**3:.2f} GB")
    print(f"Total: {total / 1024**3:.2f} GB")
    print(f"Asignada: {torch.cuda.memory_allocated() / 1024**3:.2f} GB")
    print(f"Reservada: {torch.cuda.memory_reserved() / 1024**3:.2f} GB")

mostrar_memoria()
```

```python
if torch.cuda.is_available():
    bloque = torch.randn(4096, 4096, device="cuda")
    mostrar_memoria()
else:
    bloque = torch.randn(1024, 1024)
    print("Tensor creado en CPU")
```

```python
del bloque
if torch.cuda.is_available():
    torch.cuda.empty_cache()
mostrar_memoria()
```

## 📖 Concepto clave

- **Asignada:** memoria usada por tensores vivos.
- **Reservada:** memoria que PyTorch conserva para reutilizar.
- `empty_cache()` no elimina tensores que todavía tienen referencias.

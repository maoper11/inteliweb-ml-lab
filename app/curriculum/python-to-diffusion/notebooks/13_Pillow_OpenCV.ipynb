# 13 · Procesamiento de imágenes con Pillow y OpenCV

**Objetivo:** crear, guardar, cargar, redimensionar, convertir y filtrar imágenes.

```python
from pathlib import Path
from PIL import Image, ImageDraw
import numpy as np
import cv2
import matplotlib.pyplot as plt

output_dir = Path("outputs")
output_dir.mkdir(exist_ok=True)
```

```python
imagen = Image.new("RGB", (512, 512), (30, 40, 70))
draw = ImageDraw.Draw(imagen)
draw.ellipse((120, 100, 392, 372), fill=(230, 120, 90))
draw.rectangle((190, 210, 322, 260), fill=(255, 255, 255))
imagen
```

```python
ruta = output_dir / "imagen_practica.png"
imagen.save(ruta)
print(ruta.resolve())
```

```python
array_rgb = np.array(imagen)
array_bgr = cv2.cvtColor(array_rgb, cv2.COLOR_RGB2BGR)
gris = cv2.cvtColor(array_bgr, cv2.COLOR_BGR2GRAY)
bordes = cv2.Canny(gris, 50, 150)
print("RGB:", array_rgb.shape)
print("Gris:", gris.shape)
print("Bordes:", bordes.shape)
```

```python
fig, axes = plt.subplots(1, 3, figsize=(12, 4))
axes[0].imshow(array_rgb)
axes[0].set_title("RGB")
axes[1].imshow(gris, cmap="gray")
axes[1].set_title("Escala de grises")
axes[2].imshow(bordes, cmap="gray")
axes[2].set_title("Bordes")
for ax in axes:
    ax.axis("off")
plt.show()
```

## 🧪 Práctica

Cambia los umbrales de `Canny(50, 150)` y observa cuántos bordes aparecen.
